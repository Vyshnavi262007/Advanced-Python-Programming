```python
import logging
import tracemalloc


# Configure logging
logging.basicConfig(
    filename="exam.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)


class Question:
    def __init__(self, number, question, options, answer):
        self.number = number
        self.question = question
        self.options = options
        self.answer = answer


class Student:
    def __init__(self, name):
        self.name = name
        self.answers = []


class Exam:
    def __init__(self, questions):
        self.questions = questions

    def conduct_exam(self, student):

        for q in self.questions:

            print("\n", q.number, ".", q.question)

            for option in q.options:
                print(option)

            try:
                answer = input("Enter your answer (A/B/C/D): ").upper()

                assert answer in ["A", "B", "C", "D"], \
                    "Please enter only A, B, C or D"

                student.answers.append(answer)

                logging.info("Answer recorded for Question %d", q.number)

            except AssertionError as e:
                print("Invalid input:", e)

                logging.error(
                    "Invalid answer for Question %d", q.number
                )

                student.answers.append("INVALID")

    def calculate_score(self, student):

        score = 0

        for i in range(len(self.questions)):

            if student.answers[i] == "INVALID":
                continue

            if student.answers[i] == self.questions[i].answer:
                score += 1
            else:
                score -= 0.25

        return score


if __name__ == "__main__":

    logging.info("Exam started")

    # Start memory profiling
    tracemalloc.start()

    try:
        name = input("Enter student name: ")

        assert name.strip() != "", "Student name cannot be empty"

        student = Student(name)

        logging.info("Student registered: %s", name)

    except AssertionError as e:
        print("Error:", e)
        logging.error("Invalid student name")
        exit()

    # Question set
    questions = [

        Question(
            1,
            "What is Python?",
            [
                "A. Programming Language",
                "B. Operating System",
                "C. Database",
                "D. Browser"
            ],
            "A"
        ),

        Question(
            2,
            "Which keyword is used to define a function in Python?",
            [
                "A. function",
                "B. def",
                "C. fun",
                "D. define"
            ],
            "B"
        ),

        Question(
            3,
            "Which module is used for logging in Python?",
            [
                "A. os",
                "B. sys",
                "C. logging",
                "D. math"
            ],
            "C"
        ),

        Question(
            4,
            "Which module is used for memory profiling?",
            [
                "A. random",
                "B. os",
                "C. sys",
                "D. tracemalloc"
            ],
            "D"
        ),

        Question(
            5,
            "Which keyword is used to handle exceptions?",
            [
                "A. try",
                "B. catch",
                "C. error",
                "D. exception"
            ],
            "A"
        )
    ]

    # Create exam
    exam = Exam(questions)

    # Conduct exam
    exam.conduct_exam(student)

    # Calculate score
    score = exam.calculate_score(student)

    # Get memory usage
    current, peak = tracemalloc.get_traced_memory()

    # Display result
    print("\n" + "=" * 40)
    print("EXAMINATION RESULT")
    print("=" * 40)

    print("Student Name :", student.name)
    print("Score        :", score)
    print("Total Marks  :", len(questions))

    if score >= 3:
        print("Result       : PASS")
    else:
        print("Result       : FAIL")

    print("\nMemory Usage")
    print("Current      :", current, "bytes")
    print("Peak         :", peak, "bytes")

    # Logging
    logging.info(
        "Exam completed. Student: %s, Score: %.2f",
        student.name,
        score
    )

    tracemalloc.stop()

    print("\nExam completed successfully.")
    print("Log file created: exam.log")
```
