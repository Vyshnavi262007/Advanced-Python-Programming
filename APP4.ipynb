```python
from functools import reduce
```

```python
purchases = [
    {"id": 101, "name": "Rahul", "product": "Laptop", "price": 50000, "quantity": 1},
    {"id": 102, "name": "Priya", "product": "Phone", "price": 25000, "quantity": 2},
    {"id": 103, "name": "Arjun", "product": "Headphones", "price": 2000, "quantity": 3},
    {"id": 104, "name": "Sneha", "product": "Tablet", "price": 15000, "quantity": 1},
    {"id": 105, "name": "Kiran", "product": "Watch", "price": 5000, "quantity": 2}
]

purchases
```

```python
calculate_amount = lambda p: p["price"] * p["quantity"]

calculate_amount(purchases[0])
```

```python
processed = list(map(
    lambda p: {**p, "total": p["price"] * p["quantity"]},
    purchases
))

processed
```

```python
for p in processed:
    print(
        p["id"],
        p["name"],
        p["product"],
        "Total =", p["total"]
    )
```

```python
limit = 20000

high_value = list(filter(
    lambda p: p["total"] > limit,
    processed
))

high_value
```

```python
for p in high_value:
    print(
        p["name"],
        p["product"],
        "Total =", p["total"]
    )
```

```python
total_sales = reduce(
    lambda total, p: total + p["total"],
    processed,
    0
)

total_sales
```

```python
average = total_sales / len(processed)

print("Average Purchase Value =", average)
```

```python
highest = reduce(
    lambda a, b: a if a["total"] > b["total"] else b,
    processed
)

highest
```

```python
print("Highest-Value Purchase")
print("Customer:", highest["name"])
print("Product:", highest["product"])
print("Amount:", highest["total"])
```
