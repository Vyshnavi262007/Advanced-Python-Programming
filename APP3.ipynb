```python
class Vehicle:
    def __init__(self, vehicle_id, model, category, daily_rate, hourly_rate, km_rate):
        self.vehicle_id = vehicle_id
        self.model = model
        self.category = category
        self.daily_rate = daily_rate
        self.hourly_rate = hourly_rate
        self.km_rate = km_rate

    def calculate_rent(self, duration, duration_type, km):
        if duration_type == "days":
            base_rent = self.daily_rate * duration
        else:
            base_rent = self.hourly_rate * duration

        km_charge = self.km_rate * km

        return base_rent + km_charge

    def __str__(self):
        return f"{self.category}: {self.model} (ID: {self.vehicle_id})"

    def __repr__(self):
        return f"Vehicle('{self.vehicle_id}', '{self.model}')"

    def __len__(self):
        return len(self.model)

    def __del__(self):
        print(f"{self.category} {self.vehicle_id} deleted.")
```

```python
class Car(Vehicle):
    def __init__(self, vehicle_id, model):
        super().__init__(
            vehicle_id,
            model,
            "Car",
            daily_rate=2000,
            hourly_rate=150,
            km_rate=15
        )
```

```python
class Bike(Vehicle):
    def __init__(self, vehicle_id, model):
        super().__init__(
            vehicle_id,
            model,
            "Bike",
            daily_rate=800,
            hourly_rate=80,
            km_rate=5
        )
```

```python
class ElectricScooter(Vehicle):
    def __init__(self, vehicle_id, model):
        super().__init__(
            vehicle_id,
            model,
            "Electric Scooter",
            daily_rate=600,
            hourly_rate=60,
            km_rate=4
        )
```

```python
class Rental:
    def __init__(self, customer_name, vehicle, duration, duration_type, km):
        self.customer_name = customer_name
        self.vehicle = vehicle
        self.duration = duration
        self.duration_type = duration_type
        self.km = km

    def calculate_rent(self):
        return self.vehicle.calculate_rent(
            self.duration,
            self.duration_type,
            self.km
        )

    def __str__(self):
        return (
            f"\n----- RENTAL RECEIPT -----\n"
            f"Customer       : {self.customer_name}\n"
            f"Vehicle        : {self.vehicle.model}\n"
            f"Category       : {self.vehicle.category}\n"
            f"Duration       : {self.duration} {self.duration_type}\n"
            f"Kilometers     : {self.km} km\n"
            f"Total Rent     : ₹{self.calculate_rent():.2f}\n"
            f"--------------------------"
        )

    def __repr__(self):
        return f"Rental('{self.customer_name}', '{self.vehicle.model}')"

    def __len__(self):
        return self.duration

    def __del__(self):
        print(f"Rental of {self.customer_name} deleted.")
```
