```python
import asyncio
import multiprocessing
import time


# Employee class
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary


# Payroll Processor class
class PayrollProcessor:

    # CPU-intensive salary calculation
    @staticmethod
    def calculate_salary(employee):
        salary = employee.salary

        # Simulating CPU-intensive calculation
        total = 0
        for i in range(1000000):
            total += i % 10

        tax = salary * 0.10
        bonus = salary * 0.05
        deduction = salary * 0.02

        net_salary = salary + bonus - tax - deduction

        return {
            "id": employee.emp_id,
            "name": employee.name,
            "basic_salary": salary,
            "tax": tax,
            "bonus": bonus,
            "deduction": deduction,
            "net_salary": net_salary
        }


# Async function to simulate retrieving employee information
async def get_employee_info(employee):
    await asyncio.sleep(1)

    return {
        "id": employee.emp_id,
        "name": employee.name,
        "department": "IT",
        "attendance": "Present"
    }


# Async function to retrieve information for all employees
async def get_all_employee_info(employees):
    tasks = []

    for employee in employees:
        tasks.append(get_employee_info(employee))

    results = await asyncio.gather(*tasks)

    return results


# Main function
if __name__ == "__main__":

    employees = [
        Employee(101, "Alice", 50000),
        Employee(102, "Bob", 60000),
        Employee(103, "Charlie", 55000),
        Employee(104, "David", 70000)
    ]

    print("Retrieving employee information asynchronously...\n")

    # AsyncIO part
    # Use await directly in environments where an event loop is already running (like Colab)
    employee_info = await get_all_employee_info(employees)

    print("Employee Information:")
    for info in employee_info:
        print(info)

    print("\nProcessing payroll using multiprocessing...\n")

    # Multiprocessing Pool
    with multiprocessing.Pool() as pool:
        payroll_results = pool.map(
            PayrollProcessor.calculate_salary,
            employees
        )

    print("Final Payroll Report")
    print("-" * 70)

    for payroll in payroll_results:
        print("Employee ID :", payroll["id"])
        print("Name        :", payroll["name"])
        print("Basic Salary:", payroll["basic_salary"])
        print("Tax         :", payroll["tax"])
        print("Bonus       :", payroll["bonus"])
        print("Deduction   :", payroll["deduction"])
        print("Net Salary  :", payroll["net_salary"])
        print("-" * 70)
```
