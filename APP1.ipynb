```python
class Login:
    def __init__(self):
        self.users = {}

    def register(self, email, password):
        self.users[email] = password
        print("Registration Successful")

    def login(self, email, password):
        if email in self.users and self.users[email] == password:
            print("Login Successful")
            return True
        else:
            print("Invalid Credentials")
            return False
```

```python
class User:
    def __init__(self, user_id, name, phone):
        self.user_id = user_id
        self.name = name
        self.phone = phone

    def display_user(self):
        print("User ID:", self.user_id)
        print("Name:", self.name)
        print("Phone:", self.phone)
```

```python
class Customer(User):

    def __init__(self, user_id, name, phone, address):
        super().__init__(user_id, name, phone)
        self.address = address

    def place_order(self, restaurant, items):
        print(f"{self.name} placed order at {restaurant.name}")
```

```python
class PremiumCustomer(Customer):

    def calculate_discount(self, amount):
        discount = amount * 0.20
        return discount
```

```python
class Restaurant:

    def __init__(self, restaurant_id, name, veg_menu, nonveg_menu):
        self.restaurant_id = restaurant_id
        self.name = name
        self.veg_menu = veg_menu
        self.nonveg_menu = nonveg_menu

    def display_menu(self):
        print(f"\n----- {self.name} -----")

        print("\nVeg Menu")
        for item, price in self.veg_menu.items():
            print(f"{item} - ₹{price}")

        print("\nNon-Veg Menu")
        for item, price in self.nonveg_menu.items():
            print(f"{item} - ₹{price}")
```

```python
print("\n========== ZOMATO ==========")

print("\nAvailable Restaurants:\n")

for restaurant in restaurants:
    print(f"{restaurant.restaurant_id}. {restaurant.name}")

choice = int(input("\nEnter Restaurant Number: "))

selected_restaurant = restaurants[choice - 1]

selected_restaurant.display_menu()
```

```python
food_type = input("\nDo you want Veg or Non-Veg? ").lower()

if food_type == "veg":
    menu = selected_restaurant.veg_menu
elif food_type == "nonveg" or food_type == "non-veg":
    menu = selected_restaurant.nonveg_menu
else:
    print("Invalid Choice")
    menu = {}
```

```python
cart = {}
total = 0

while True:

    item = input("\nEnter Item Name (or 'done' to finish): ")

    if item.lower() == "done":
        break

    if item in menu:

        qty = int(input("Enter Quantity: "))

        cart[item] = qty
        total += menu[item] * qty

    else:
        print("Item not found!")

print("\n=========== BILL ===========")

for item, qty in cart.items():
    print(f"{item:25} x {qty} = ₹{menu[item] * qty}")

print("----------------------------")
print("Total Amount = ₹", total)
```

```python
class Order:

    def __init__(self, customer, restaurant):
        self.customer = customer
        self.restaurant = restaurant
        self.items = []
        self.total = 0

    def add_item(self, item, quantity):
        if item in self.restaurant.menu:
            price = self.restaurant.menu[item]
            self.items.append((item, quantity, price))
            self.total += quantity * price

    def generate_bill(self):

        if isinstance(self.customer, PremiumCustomer):
            discount = self.customer.calculate_discount(self.total)
        else:
            discount = 0

        final_amount = self.total - discount

        print("\n------ BILL ------")

        for item, qty, price in self.items:
            print(item, qty, price)

        print("Total:", self.total)
        print("Discount:", discount)
        print("Final Amount:", final_amount)

    def display_result(self):
        print("\nOrder Successful")
        print("Customer:", self.customer.name)
        print("Restaurant:", self.restaurant.name)
```

```python
# ================= LOGIN CLASS =================

class Login:

    def __init__(self):
        self.users = {}

    def register(self):
        print("\n========== REGISTER ==========")

        name = input("Enter Name: ")
        email = input("Enter Email: ")
        phone = input("Enter Phone Number: ")
        password = input("Create Password: ")

        if email in self.users:
            print("User already exists!")
        else:
            self.users[email] = {
                "name": name,
                "phone": phone,
                "password": password
            }
            print("Registration Successful!")

    def login(self):
        print("\n========== LOGIN ==========")

        email = input("Enter Email: ")
        password = input("Enter Password: ")

        if email in self.users:
            if self.users[email]["password"] == password:
                print(f"\nWelcome {self.users[email]['name']}!")
                return True
            else:
                print("Incorrect Password!")
                return False
        else:
            print("User not found!")
            return False
```

```python
login = Login()

while True:

    print("\n========== ZOMATO ==========")
    print("1. Register")
    print("2. Login")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        login.register()

    elif choice == 2:
        if login.login():
            print("Login Successful!")
            break

    elif choice == 3:
        print("Thank you for using Zomato!")
        break

    else:
        print("Invalid Choice!")
```
