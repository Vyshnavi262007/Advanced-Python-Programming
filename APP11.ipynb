```python
import pandas as pd
```

```python
data = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Rahul", "Priya", "Arjun", "Sneha"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE"],
    "CGPA": [8.5, 7.2, None, 8.8, 7.5],
    "Skills": ["Python", "Java", "Python", None, "SQL"]
}

df = pd.DataFrame(data)
df.to_csv("candidates.csv", index=False)
```

```python
df = pd.read_csv("candidates.csv")

print(df)
```

```python
print("Missing Values:")
print(df.isnull().sum())

df["CGPA"] = df["CGPA"].fillna(df["CGPA"].mean())
df["Skills"] = df["Skills"].fillna("Not Specified")

print("\nAfter Cleaning:")
print(df)
```

```python
df["Status"] = df["CGPA"].apply(
    lambda x: "Eligible" if x >= 7.0 else "Not Eligible"
)

print("Validated Candidates:")
print(df[["ID", "Name", "CGPA", "Status"]])
```

```python
report = df.groupby("Department").agg(
    Candidates=("ID", "count"),
    Average_CGPA=("CGPA", "mean")
)

print("Department Report:")
print(report)
```

```python
company = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Company": ["TCS", "Infosys", "Google", "Wipro", "TCS"]
})

merged = pd.merge(df, company, on="ID")

print("Merged Data:")
print(merged)
```

```python
pivot = pd.pivot_table(
    merged,
    values="ID",
    index="Department",
    columns="Company",
    aggfunc="count",
    fill_value=0
)

print("Pivot Table:")
print(pivot)
```

```python
print("\n===== CANDIDATE SUMMARY =====")
print("Total Candidates:", len(df))
print("Eligible Candidates:", (df["Status"] == "Eligible").sum())
print("Average CGPA:", round(df["CGPA"].mean(), 2))
```
