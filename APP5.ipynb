```python
class ElectricityReading:
    def __init__(self, meter_id, reading):
        self.meter_id = meter_id
        self.reading = reading

    def display(self):
        print("Meter ID:", self.meter_id,
              "| Consumption:", self.reading, "kWh")
```

```python
def generate_readings(readings):
    for reading in readings:
        yield reading
```

```python
class ConsumptionMonitor:

    def __init__(self, readings):
        self.readings = readings
        self.index = 0

    def __iter__(self):
        self.index = 0
        return self

    def __next__(self):

        if self.index >= len(self.readings):
            raise StopIteration

        reading = self.readings[self.index]
        self.index += 1

        return reading
```

```python
readings = [
    ElectricityReading("M001", 120),
    ElectricityReading("M002", 250),
    ElectricityReading("M003", 180),
    ElectricityReading("M004", 350),
    ElectricityReading("M005", 150)
]
```

```python
generator = generate_readings(readings)

print("----- ELECTRICITY CONSUMPTION READINGS -----")
```

```python
total = 0
count = 0
threshold = 200
highest = None

for reading in generator:

    reading.display()

    total = total + reading.reading
    count = count + 1

    if reading.reading > threshold:
        print("WARNING: High consumption!")

    if highest is None or reading.reading > highest.reading:
        highest = reading
```

```python
average = total / count

print("\n----- CONSUMPTION SUMMARY -----")

print("Total Consumption:", total, "kWh")
print("Average Consumption:", average, "kWh")
print("Threshold:", threshold, "kWh")
```

```python
print("\nHighest Consumption:")
highest.display()
```

```python
print("\n----- ITERATOR DEMONSTRATION -----")

monitor = ConsumptionMonitor(readings)

for reading in monitor:
    reading.display()
```
