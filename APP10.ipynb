```python
import numpy as np
```

```python
locations = ["Main Road", "City Center", "Residential Area", "Highway Junction"]
time_slots = ["Morning", "Afternoon", "Evening"]

print("Locations:", locations)
print("Time Slots:", time_slots)
print("Time Slots:", time_slots)
```

```python
traffic_data = [
    [120, 150, 180],
    [200, 220, 250],
    [80, 100, 90],
    [160, 190, 210]
]
```

```python
traffic_array = np.array(traffic_data)

print("Traffic Data:")
print(traffic_array)
```

```python
traffic_array = traffic_array.reshape(4, 3)

print("Reshaped Traffic Data:")
print(traffic_array)
```

```python
total_traffic = np.sum(traffic_array)

print("Total Traffic:", total_traffic)
```

```python
global_threshold = np.mean(traffic_array)

print("Global Traffic Threshold:", global_threshold)
```

```python
traffic_std = np.std(traffic_array)

print("Standard Deviation:", traffic_std)
```

```python
total_traffic = np.sum(traffic_array, axis=1)
avg_traffic = np.mean(traffic_array, axis=1)
variation = np.std(traffic_array, axis=1)
```

```python
print("\n========== SMART TRAFFIC SUMMARY ==========")

print("Global Traffic Threshold:", global_threshold)

print("\nLocation\t\tTotal Traffic\tAvg/Hour\tVariation (Std)")

for i in range(len(locations)):
    print(f"{locations[i]:20}\t{total_traffic[i]:12}\t{avg_traffic[i]:8.2f}\t{variation[i]:8.2f}")
```

```python
print("\n========== TRAFFIC ALERT ==========")

high_traffic = avg_traffic > global_threshold

for i in range(len(locations)):
    if high_traffic[i]:
        print("ALERT: High traffic detected at", locations[i])
```
