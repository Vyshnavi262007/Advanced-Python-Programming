```python
import numpy as np
```

```python
traffic_data = [
    [120, 150, 180],
    [200, 220, 250],
    [80, 100, 90],
    [160, 190, 210]
]
```

```python
traffic_array = np.array(traffic_data)

print("Traffic Data:")
print(traffic_array)
```

```python
traffic_array = traffic_array.reshape(4, 3)

print("Reshaped Traffic Data:")
print(traffic_array)
```

```python
total_traffic = np.sum(traffic_array)

print("Total Traffic:", total_traffic)
```

```python
average_traffic = np.mean(traffic_array)

print("Average Traffic:", average_traffic)
```

```python
traffic_std = np.std(traffic_array)

print("Standard Deviation:", traffic_std)
```

```python
location_total = np.sum(traffic_array, axis=1)
location_average = np.mean(traffic_array, axis=1)

print("Total traffic at each location:")
print(location_total)

print("\nAverage traffic at each location:")
print(location_average)
```

```python
high_traffic = location_average >= 180

print("High-Traffic Locations:")
print(high_traffic)
```

```python
most_congested = np.argmax(location_average)

print("Most Congested Location: Location", most_congested + 1)
print("Average Traffic:", location_average[most_congested])
```

```python
time_average = np.mean(traffic_array, axis=0)

print("Average Traffic for Each Time Slot:")
print(time_average)
```

```python
# Define categories based on the high_traffic boolean array
categories = ["High Traffic" if is_high else "Normal Traffic" for is_high in high_traffic]

print("----- SMART TRAFFIC DATA SUMMARY -----")

print("Total Traffic:", total_traffic)
print("Overall Average Traffic:", average_traffic)
print("Standard Deviation:", traffic_std)

for i in range(4):
    print(
        "Location", i + 1,
        "- Total:", location_total[i],
        "- Average:", location_average[i],
        "- Category:", categories[i]
    )

print("Most Congested Location:", most_congested + 1)
```
