```python
class BankAccount:

    def __init__(self, name, balance, pin):
        self.name = name
        self.__balance = balance
        self.__pin = pin

    def get_balance(self):
        return self.__balance

    def set_pin(self, new_pin):
        if len(str(new_pin)) == 4:
            self.__pin = new_pin
            print("PIN changed successfully")
        else:
            print("PIN must contain exactly 4 digits")
    # Operator overloading +
    def __add__(self, other):
        return self.__balance + other.__balance

    # Operator overloading >
    def __gt__(self, other):
        return self.__balance > other.__balance
```

```python
account = BankAccount("Vyshnavi", 10000, 1234)

print(account.get_balance())

account.set_pin(5678)
```

```python
class SavingsAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.04


class CurrentAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.02


class FixedDepositAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.07
```

```python
savings = SavingsAccount("Vyshnavi", 10000, 1234)
current = CurrentAccount("Rahul", 10000, 2345)
fixed = FixedDepositAccount("Anu", 10000, 3456)

print("Savings Account Interest:", savings.calculate_interest())
print("Current Account Interest:", current.calculate_interest())
print("Fixed Deposit Interest:", fixed.calculate_interest())
```
